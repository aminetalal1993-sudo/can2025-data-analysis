# 🏆 Analyse de Données : Coupe d'Afrique des Nations 2025 (Maroc)
**Auteur :** Amine | **Outils :** Python (Pandas, SQLite3, Matplotlib)  
Ce notebook présente les **4 volets analytiques** de notre étude sur la CAN 2025 :
1. **Audit & Nettoyage des données (Data Cleaning)**
2. **Évolution Historique de la CAN (1957 - 2025)**
3. **Bilan Collectif des 24 Équipes & Focus Maroc 🇲🇦**
4. **Scouting & Détection des Pépites U23**

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

conn = sqlite3.connect('../data/can2025.db')
df_teams = pd.read_sql_query('SELECT * FROM teams', conn)
df_matches = pd.read_sql_query('SELECT * FROM matches', conn)
df_players = pd.read_sql_query('SELECT * FROM players', conn)
print(f'Équipes : {len(df_teams)} | Matchs : {len(df_matches)} | Joueurs clés suivis : {len(df_players)}')
```

## 🇲🇦 Focus Spécial : Performance des Lions de l'Atlas (Maroc)
Analysons les joueurs marocains les mieux notés et leur apport offensif/défensif.

```python
df_maroc = df_players[df_players['team_id'] == 'MAR'].sort_values('rating', ascending=False)
df_maroc[['player_name', 'position', 'club', 'goals', 'assists', 'key_passes', 'progressive_carries', 'duels_won_pct', 'rating']]
```

## 🔍 Cellule de Recrutement : Les Pépites U23 sous-cotées (<= 20 M€)
Quels jeunes joueurs (23 ans ou moins) ont affiché des performances d'élite tout en ayant une valeur marchande accessible ?

```python
pepites = df_players[(df_players['age'] <= 23) & (df_players['market_value_m_eur'] <= 20.0)].sort_values('rating', ascending=False)
pepites[['player_name', 'team_name', 'age', 'position', 'club', 'market_value_m_eur', 'goals', 'assists', 'rating']]
```
